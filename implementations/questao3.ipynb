## Calcular Numericamente o vão espectral (item 2)

```python
import numpy as np

def spectral_gap(n, p):
    P = np.zeros((n, n))
    
    for i in range(n):
        P[i,0] = 1-p
        if i+1 < n: 
            P[i,i+1] = p

    P[n-1,n-1] = p

    eigvals = np.linalg.eigvals(P)
    
    mods = np.abs(eigvals)
    
    mods_sorted = np.sort(mods)[::-1]
    
    spectral_gap = 1 - mods_sorted[1]
    
    return spectral_gap, P
```

```python
# Exemplo de uso
n = 16

for p in [0.25, 0.5, 0.75]:
    gap, P = spectral_gap(n, p)
    print(f"Spectral gap para n={n}, p={p}: {gap}")
```

## Calcular Numericamente a distribuição estacionária (item 3)
(usando autovalores e autovetores)

```python
import numpy as np

def stationary_distribution(n, p):

    # matriz de transição
    P = np.zeros((n,n))

    for i in range(n):
        P[i,0] = 1-p

    for i in range(n-1):
        P[i,i+1] = p

    P[n-1,n-1] = p

    eigvals, eigvecs = np.linalg.eig(P.T)
    
    idx = np.argmin(np.abs(eigvals - 1))

    pi = np.real(eigvecs[:,idx])


    pi = pi / np.sum(pi)

    return pi
```

```python
n = 16

for p in [0.25, 0.5, 0.75]:
    pi = stationary_distribution(n, p)

    print(f"\nDistribuição estacionária para n={n}, p={p}")
    print("-" * 50)

    for i, prob in enumerate(pi, start=1):
        print(f"Estado {i:2d}: {prob:.2e}")

    min_prob = np.min(pi)
    min_states = np.where(np.isclose(pi, min_prob, rtol=1e-6))[0] + 1

    print("-" * 50)
    if len(min_states) == 1:
        print(f"Estado de menor probabilidade: {min_states[0]} (π = {min_prob:.2e})")
    else:
        print(f"Estados de menor probabilidade (empate): {list(min_states)} (π = {min_prob:.2e})")
```
