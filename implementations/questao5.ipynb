## Simulador Cadeia de Markov (item 1)

```python
import numpy as np

def simulate_chain(p, T, start=(1,1), seed=None):
    """
    Simula a cadeia de Markov no reticulado 2D.

    Estados: (i,j), com i,j >= 1

    Probabilidades:
        Norte  -> (i, j+1) com prob. p/2
        Leste  -> (i+1, j) com prob. p/2
        Sul    -> (i, j-1) com prob. (1-p)/2
        Oeste  -> (i-1, j) com prob. (1-p)/2

    Nas bordas utiliza self-loop.
    """

    if seed is not None:
        np.random.seed(seed)

    state = list(start)

    trajectory = [tuple(state)]

    for _ in range(T):

        i, j = state

        r = np.random.rand()

        # Norte
        if r < p/2:
            j += 1

        # Leste
        elif r < p:
            i += 1

        # Sul
        elif r < p + (1-p)/2:

            if j > 1:
                j -= 1
            # senão self-loop

        # Oeste
        else:

            if i > 1:
                i -= 1
            # senão self-loop

        state = [i, j]

        trajectory.append(tuple(state))

    return trajectory
```

```python
# Exemplo de uso
T = 100
p = 0.35

traj = simulate_chain(p, T, seed=15)

print("Primeiros estados visitados:")
for s in traj[:20]:
    print(s)
```

## Calculo do tempo de retorno e distribuição estacionária (item 2)

```python
def estimate_pi_origin(p, T=1000000):

    traj = simulate_chain(p, T)

    return_times = []

    last_visit = None

    for t, state in enumerate(traj):

        if state == (1,1):

            if last_visit is not None:

                return_times.append(t - last_visit)

            last_visit = t

    mean_return = np.mean(return_times)

    pi_est = 1 / mean_return

    return mean_return, pi_est
```

```python
# Exemplo
for p in [0.25, 0.35, 0.45]:

    mean_return, pi_est = estimate_pi_origin(p)

    print(f"\np = {p}")
    print(f"Tempo médio de retorno: {mean_return:.4f}")
    print(f"Estimativa de pi_(1,1): {pi_est:.6f}")
```

## Estimar Distancia de Manhattan entre origem e X_t (item 3)

```python
def manhattan_distance(state):
    i, j = state

    # distância até a origem (1,1)
    return abs(i-1) + abs(j-1)
```

```python
def estimate_d(p, t, num_simulations=10000):

    distances = []

    for _ in range(num_simulations):

        traj = simulate_chain(p, t)

        final_state = traj[-1]

        distances.append(manhattan_distance(final_state))

    return np.mean(distances)
```

```python
times = [10, 100, 1000]

for p in [0.25, 0.5, 0.75]:

    print(f"\np = {p}")
    print("-" * 40)

    for t in times:

        d = estimate_d(p, t)

        print(f"d({t}) ≈ {d:.4f}")
```
